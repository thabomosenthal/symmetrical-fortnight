## Part 2: 💬 Building a WhatsApp Bot

### Simple Echo Bot

````python
# Simple WhatsApp bot with Twilio

ECHO_BOT_CODE = '''
from flask import Flask, request
from twilio.twiml.messaging_response import MessagingResponse
from twilio.rest import Client
import os

app = Flask(__name__)

# Twilio credentials (use environment variables!)
account_sid = os.getenv('TWILIO_ACCOUNT_SID')
auth_token = os.getenv('TWILIO_AUTH_TOKEN')
client = Client(account_sid, auth_token)

@app.route('/webhook', methods=['POST'])
def webhook():
    """Handle incoming WhatsApp messages."""

    # Get incoming message
    incoming_msg = request.values.get('Body', '').strip()
    sender = request.values.get('From', '')

    # Create response
    resp = MessagingResponse()
    msg = resp.message()

    # Echo back
    msg.body(f'You said: {incoming_msg}')

    return str(resp)

if __name__ == '__main__':
    app.run(port=5000, debug=True)
'''

print('🤖 Simple Echo Bot:')
print(ECHO_BOT_CODE)
print('\n💡 To test:')
print('  1. Run: python bot.py')
print('  2. Expose with ngrok: ngrok http 5000')
print('  3. Set webhook URL in Twilio console')
print('  4. Send WhatsApp message to sandbox number')
````

### Intelligent Bot with Context

````python
# WhatsApp bot with session management

INTELLIGENT_BOT_CODE = '''
from flask import Flask, request, session
from twilio.twiml.messaging_response import MessagingResponse
import os
from datetime import datetime

app = Flask(__name__)
app.secret_key = os.getenv('SECRET_KEY', 'your-secret-key')

# Simple in-memory session store (use Redis in production!)
user_sessions = {}

def get_session(phone_number):
    """Get or create user session."""
    if phone_number not in user_sessions:
        user_sessions[phone_number] = {
            'messages': [],
            'context': {},
            'created_at': datetime.now(),
            'last_active': datetime.now(),
        }
    user_sessions[phone_number]['last_active'] = datetime.now()
    return user_sessions[phone_number]

def generate_response(user_message, user_session):
    """Generate AI response based on message and context."""
    # This is where you integrate your LLM!
    # For now, simple pattern matching

    message_lower = user_message.lower()

    # Greetings
    if any(word in message_lower for word in ['hi', 'hello', 'hey']):
        name = user_session['context'].get('name', 'there')
        return f"Hello {name}! How can I help you today?"

    # Name capture
    if 'my name is' in message_lower:
        name = user_message.split('my name is')[-1].strip()
        user_session['context']['name'] = name
        return f"Nice to meet you, {name}! I'll remember that."

    # Help
    if 'help' in message_lower:
        return """I can help with:
• Answering questions
• Identifying images (send a photo)
• Remembering context
• Multi-language support

What would you like to do?"""

    # Default
    return f"I received: {user_message}. I'm learning to respond better!"

@app.route('/webhook', methods=['POST'])
def webhook():
    """Handle incoming WhatsApp messages."""

    # Get message details
    incoming_msg = request.values.get('Body', '').strip()
    sender = request.values.get('From', '')
    media_url = request.values.get('MediaUrl0', None)  # For images

    # Get user session
    user_session = get_session(sender)

    # Add to history
    user_session['messages'].append({
        'role': 'user',
        'content': incoming_msg,
        'media': media_url,
        'timestamp': datetime.now(),
    })

    # Generate response
    if media_url:
        response_text = "I received your image! Image analysis coming soon."
    else:
        response_text = generate_response(incoming_msg, user_session)

    # Add to history
    user_session['messages'].append({
        'role': 'assistant',
        'content': response_text,
        'timestamp': datetime.now(),
    })

    # Send response
    resp = MessagingResponse()
    msg = resp.message()
    msg.body(response_text)

    return str(resp)

if __name__ == '__main__':
    app.run(port=5000, debug=True)
'''

print('🧠 Intelligent WhatsApp Bot:')
print(INTELLIGENT_BOT_CODE[:500] + '...\n[Full code above]')
````

## Part 3: 🔗 Integrating Your Digital Twin

### Converting Web App to WhatsApp Bot

````python
# Integration pattern

INTEGRATION_PATTERN = '''
from flask import Flask, request
from twilio.twiml.messaging_response import MessagingResponse
import anthropic  # Your LLM
import os

app = Flask(__name__)

# Initialize your digital twin components
# (Import from your Week 1-3 code)
from my_digital_twin import DigitalTwin

# Create client
anthropic_client = anthropic.Anthropic(
    api_key=os.getenv('ANTHROPIC_API_KEY')
)

# Session storage
user_twins = {}

def get_twin(phone_number):
    """Get or create digital twin for user."""
    if phone_number not in user_twins:
        user_twins[phone_number] = DigitalTwin(
            name=f"Twin_{phone_number[-4:]}",
            client=anthropic_client
        )
    return user_twins[phone_number]

@app.route('/webhook', methods=['POST'])
def webhook():
    """Handle incoming messages."""
    incoming_msg = request.values.get('Body', '')
    sender = request.values.get('From', '')

    # Get user's twin
    twin = get_twin(sender)

    # Generate response using your full digital twin
    response_text = twin.chat(incoming_msg)

    # Send via WhatsApp
    resp = MessagingResponse()
    resp.message(response_text)

    return str(resp)

if __name__ == '__main__':
    app.run(port=5000)
'''

print('🔗 Integration Pattern:')
print(INTEGRATION_PATTERN)
print('\n✨ Key idea: Your web app logic works in WhatsApp too!')
print('   Just swap the interface (Streamlit → WhatsApp)')
````

### Handling Media (Images, Audio)

````python
# Media handling example

MEDIA_HANDLING_CODE = '''
import requests
from PIL import Image
from io import BytesIO
import base64

@app.route('/webhook', methods=['POST'])
def webhook():
    incoming_msg = request.values.get('Body', '')
    sender = request.values.get('From', '')

    # Check for media
    num_media = int(request.values.get('NumMedia', 0))

    if num_media > 0:
        # Get first media URL
        media_url = request.values.get('MediaUrl0')
        media_type = request.values.get('MediaContentType0')

        # Download media
        media_response = requests.get(media_url)

        if 'image' in media_type:
            # Process image
            image = Image.open(BytesIO(media_response.content))

            # Convert to base64 for AI vision API
            buffered = BytesIO()
            image.save(buffered, format="JPEG")
            img_base64 = base64.b64encode(buffered.getvalue()).decode()

            # Send to Claude with vision
            response = anthropic_client.messages.create(
                model="claude-sonnet-4-20250514",
                max_tokens=1024,
                messages=[{
                    "role": "user",
                    "content": [
                        {
                            "type": "image",
                            "source": {
                                "type": "base64",
                                "media_type": "image/jpeg",
                                "data": img_base64,
                            },
                        },
                        {
                            "type": "text",
                            "text": incoming_msg or "What's in this image?"
                        }
                    ],
                }]
            )

            response_text = response.content[0].text

        elif 'audio' in media_type:
            # Handle voice notes
            response_text = "I received your voice note! Audio transcription coming soon."

        else:
            response_text = "I received your file, but I can only process images right now."

    else:
        # Text only
        response_text = generate_text_response(incoming_msg)

    # Send response
    resp = MessagingResponse()
    resp.message(response_text)
    return str(resp)
'''

print('📸 Media Handling:')
print(MEDIA_HANDLING_CODE[:500] + '...\n[Full code above]')
print('\n✨ Your bot can now:')
print('  • Analyze images (crop diseases, receipts, etc.)')
print('  • Process voice notes (future: transcription)')
print('  • Handle documents')
````

### Performance & Cost Optimization

````python
# Cost and performance tips

OPTIMIZATION_TIPS = {
    'Rate Limiting': {
        'Why': 'Prevent abuse, control costs',
        'How': 'Limit messages per user per minute',
        'Code': '''
from collections import defaultdict
from datetime import datetime, timedelta

user_message_times = defaultdict(list)

def is_rate_limited(phone_number, limit=10, window=60):
    """Check if user exceeded rate limit."""
    now = datetime.now()
    cutoff = now - timedelta(seconds=window)

    # Remove old messages
    user_message_times[phone_number] = [
        t for t in user_message_times[phone_number]
        if t > cutoff
    ]

    # Check limit
    if len(user_message_times[phone_number]) >= limit:
        return True

    # Add current message
    user_message_times[phone_number].append(now)
    return False
''',
    },
    'Caching': {
        'Why': 'Reduce API calls, faster responses',
        'What to cache': [
            'Common questions → answers',
            'User preferences',
            'Static data (FAQs, menus)',
        ],
        'Tools': 'Redis, in-memory dict for simple cases',
    },
    'Session Cleanup': {
        'Why': 'Free memory, prevent leaks',
        'How': 'Delete sessions older than 24 hours',
        'Code': '''
import threading
import time

def cleanup_old_sessions():
    """Remove sessions older than 24 hours."""
    while True:
        now = datetime.now()
        cutoff = now - timedelta(hours=24)

        for phone, session in list(user_sessions.items()):
            if session['last_active'] < cutoff:
                del user_sessions[phone]

        time.sleep(3600)  # Run hourly

# Start cleanup thread
cleanup_thread = threading.Thread(target=cleanup_old_sessions, daemon=True)
cleanup_thread.start()
''',
    },
}

print('⚡ Performance & Cost Optimization:\n')
for tip_name, details in OPTIMIZATION_TIPS.items():
    print(f'{tip_name}:')
    if isinstance(details, dict):
        for key, value in details.items():
            if isinstance(value, list):
                print(f'  {key}:')
                for item in value:
                    print(f'    • {item}')
            else:
                print(f'  {key}: {value}'[:200])
    print()
````
